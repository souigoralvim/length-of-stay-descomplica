# Decision tree model
## 1. Importing libraries and loading data

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import (accuracy_score, classification_report, confusion_matrix)
```

```python
data = pd.read_csv(Path.cwd() / ".." / "data" / "processed" / "healthcare_dataset_processed.csv")
data_without_insurance = pd.read_csv(Path.cwd() / ".." / "data" / "processed" / "processed_dataset_without_insurance.csv")
```

## 2. Split data by train and test

### 2.1 Variables

```python
y = data['LOS Category']
X1 = data.drop(columns=['LOS Category'])
X2 = data_without_insurance.drop(columns=['LOS Category'])
```

### 2.1 Splitting

```python
train_X1, test_X1, train_y, test_y = train_test_split(X1, y, test_size=0.2, random_state=42, stratify=y) # Stratify to maintain class distribution in train and test sets
train_X2, test_X2, train_y, test_y = train_test_split(X2, y, test_size=0.2, random_state=42, stratify=y) 
```

## 3. Fit model

Fit data with insurance provider

```python
decision_tree_model = DecisionTreeClassifier(random_state=42)
decision_tree_model.fit(train_X1, train_y)
```

Fit data without insurance provider

```python
decision_tree_model_without_insurance = DecisionTreeClassifier(random_state=42)
decision_tree_model_without_insurance.fit(train_X2, train_y)
```

## 4. Predict

Predict data with insurance

```python
predictions_with_insurance = decision_tree_model.predict(test_X1)
```

Predict Data without insurance provider

```python
predictions_without_insurance = decision_tree_model_without_insurance.predict(test_X2)
```

## 5. Metrics

### 5.1 Accuracy
Accuracy indicates how many classifications were correct from all classifications. The general formula is:

$Accuracy = \frac{TP + TN}{TP + TN + FP + FN}$

Where: 
- TP = True Positive
- TN = True Negative
- FP = False Positive
- FN = False Negative

```python
accuracy_with_insurance = accuracy_score(test_y, predictions_with_insurance)
accuracy_without_insurance = accuracy_score(test_y, predictions_without_insurance)
print("Accuracy with insurance:", accuracy_with_insurance) 
print("Accuracy without insurance:", accuracy_without_insurance)
```

Accuracy with insurance: 0.7516216216216216

Accuracy without insurance: 0.7649549549549549



### 5.2 Classification Report

Classification report brings the following metrics:

$Precision: \frac{TP}{TP+FP}$

$Recall: \frac{TP}{TP+FN}$

$F1-Score: 2 \times \frac{Precision \times Recall}{Precision + Recall}$ (The F1-Score is the harmonic mean between Precision and Recall)

```python
classification_report_with_insurance = classification_report(test_y, predictions_with_insurance)
classification_report_without_insurance = classification_report(test_y, predictions_without_insurance)
print("Classification Report with insurance:")
print(classification_report_with_insurance)
print("Classification Report without insurance:")
print(classification_report_without_insurance)
```

### 5.3 Confusion Matrix

```python
confusion_matrix_with_insurance = confusion_matrix(test_y, predictions_with_insurance)
confusion_matrix_without_insurance = confusion_matrix(test_y, predictions_without_insurance)
```

```python
fig, axes = plt.subplots(1, 2, figsize=(14, 6))

# With Insurance Provider
sns.heatmap(
    confusion_matrix_with_insurance,
    annot=True,
    fmt='d',
    cmap='Blues',
    xticklabels=decision_tree_model.classes_,
    yticklabels=decision_tree_model.classes_,
    ax=axes[0]
)

axes[0].set_title('With Insurance Provider')

# Without Insurance Provider
sns.heatmap(
    confusion_matrix_without_insurance,
    annot=True,
    fmt='d',
    cmap='Blues',
    xticklabels=decision_tree_model.classes_,
    yticklabels=decision_tree_model.classes_,
    ax=axes[1]
)

axes[1].set_title('Without Insurance Provider')

plt.tight_layout()
plt.show()
```
